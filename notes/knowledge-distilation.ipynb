# Knowledge distilation

## What is knowlodge distilation (KD)?

KD is one of machine learning techniques that ___can compress model and transfer knowledge___. The main idea of knowledge distilation is to transfer the learnings (both knowldge and how to reason) of a large pre-trained model (aka. ___teacher model___) to a smaller model (aka. ___student model___) 

From [1], the concept was originated in 2006 from the paper called "Model Compression" by [Caruana](https://scholar.google.com/citations?user=B2U8EUwAAAAJ&hl=en) et al. 



## How does it work?

In knowledge distilation, the teacher model and the student model do not noeed to have the same architectures. According to Hinton et al, ___KD interprets a model's knowledge not as the strictly mathematical parameters it learns in training, but as how it generalizes to new data after that training___ (original: "a learned maooing from input vectors to output vector").

As the core concept, the KD does not only aim to train student models to mimic the teacher model's final output for a given input, but also the reasoning steps the teacher model takes to arrive at that final output (_learn both what and how_) 

The loss functions that are used in knowledge distilation are not the same as in the conventional supervised/self-supervised learning. KD used specialized type of loss function - ___distilation loss___ . that uses _discrete reasoning steps as soft targets_ for optimization
